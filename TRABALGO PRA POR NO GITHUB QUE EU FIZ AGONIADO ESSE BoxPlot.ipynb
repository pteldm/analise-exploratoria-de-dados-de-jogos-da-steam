```python
# PREÇO x DESCONTO - GRÁFICO

'''
Essa parte do código vai assimilar usando histogramas o preço dos jogos e os descontos deles.
'''                                 
Precos_Jogos = dataset[dataset['Price'] < 60]['Price'].dropna()

Descontos_Jogos = dataset[dataset['Discount'] < 60]['Discount'].dropna()

fig, ax = plt.subplots(1, 2, figsize=(14, 6)) 

# GRÁFICO 1

ax[0].hist(Precos_Jogos, bins = 30, color = 'pink', edgecolor = 'black')

ax[0].set_title('Histograma Preços dos Jogos')

ax[0].set_xlabel('Preços')

ax[0].set_ylabel('Frequência')

ax[0].grid(axis='y', linestyle='--', alpha=0.7)

# GRÁFICO 2

ax[1].hist(Descontos_Jogos, bins = 30, color = 'lightcoral', edgecolor = 'black')

ax[1].set_title('Histograma Desconto dos Jogos')

ax[1].set_xlabel('Descontos')

ax[1].set_ylabel('Frequência')

ax[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.suptitle('Análise Comparativa: Preço x Desconto', fontsize=16, y=1.02)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python

# BoxPlot para analisar a distribuição central dos preços.

'''
Para gerar esse BoxPlot precisamos calcular os Q1 e Q3 para estabelecermos um limite ligeiramente maior que 
o limite superior para não perdemos informações importantes e mantermos o BoxPlot legível.
'''
Precos_Limpos = dataset['Price'].dropna()

Q1 = Precos_Limpos.quantile(0.25)
Q3 = Precos_Limpos.quantile(0.75)
IQR = Q3 - Q1
Limite_BoxPlot_Padrao = Q3 + 1.5 * IQR
Limite_Ponderado = Q3 + 2.0 * IQR 

'''
A varíavel Limite_Ponderado é apenas limite modificado, foi necessário esasa modificação devido a quantidade grande outliers, fazendo essa
alteração do limite conseguimos eliminar outliers que atrapalhariam a construção do BoxPlot.
'''

plt.figure(figsize=(8, 10))

sns.boxplot(y=Precos_Limpos, color='lightgreen', width=0.5, fliersize=5) 

plt.ylim(bottom=0, top=Limite_Ponderado * 1.05) 

# Adiciona as linhas de referência.
plt.axhline(Limite_BoxPlot_Padrao, color='red', linestyle='--', linewidth=1,  label=f'BoxPlot Padrão ({Limite_BoxPlot_Padrao:.2f})')
plt.axhline(Limite_Ponderado, color='blue', linestyle='-', linewidth=1, label=f'Limite Ponderado ({Limite_Ponderado:.2f})')

plt.title(' Distribuição Central de Preços ')
plt.ylabel('Preço')
plt.legend()
plt.show()


'''
Conclusões sobre este BoxPlot:

A primeira que vemos ao analisarmos é que a maioria dos jogos filtrados são baratos, pois a mediana (Q2) é baixa, o que confirma que os jogos desse catálogo
tendem a ser mais acessíveis. 

A dispersão desse BoxPlot é nitidamente assimétrica, isso implica que o valor mínimo e o Q1 na parte inferior do BoxPlot são pequenas e o Q3 e o limite são 
mais longos. Isso quer dizer que, o aumento de preço dos jogos acelera SIGNIFICATIVAMENTE apenas para os jogos mais caros.
'''
```
