# Jogo com o maior número de DLC's

```python
# CONSTUCAO DO GRAFICO DOS JOGOS COM MAIOR QUANTIA DE DLC

Data_Set_Filtado = dataset.sort_values(by='DLC count', ascending=False).head(10)

Grafico_Eixo_X = Data_Set_Filtado['Name']

Grafico_Eixo_Y = Data_Set_Filtado['DLC count']

plt.figure(figsize=(14, 6))

plt.bar(Grafico_Eixo_X, Grafico_Eixo_Y, color='darkred')

plt.title('Jogos com maiores quantidades de DLC', fontsize=16)

plt.ylabel("Quantidade de DLCs", fontsize=12) 

plt.xticks(rotation=45, ha='right')


plt.tight_layout()

plt.show()

```

```python
# GRAFICO PARA FILTRAR PREÇOS DOS JOGOS DA STEAM

Filtra_Preço_Jogos = dataset[dataset['Price'] < 60].head(15)

#display(Filtra_Preço_Jogos)

Grafico_Eixo_X = Filtra_Preço_Jogos['Name']

Grafico_Eixo_Y = Filtra_Preço_Jogos['Price']

plt.figure(figsize=(18, 7))

plt.bar(Grafico_Eixo_X, Grafico_Eixo_Y, color='darkred')

plt.title('Nome dos jogos', fontsize=15)

plt.ylabel("Preços (USD)", fontsize=12) 

plt.xticks(rotation=45, ha = 'right', va = 'top',  fontsize = 12)

plt.tight_layout()

plt.show()

'''
Percebemos no código que filtramos os 15 primeiros jogos com preço abaixo de 60 dólares (USD) para analisar-mos de maneira
mais clara como este gráfico funciona. O gráfico tem um valor mínimo de 0 e máximo de 25 doláres, vemos também que o jogo
com preço mais aproximado do valor máximo é WARSAW e o jogo mais próximo do valor mínimo é Train Bandit.
'''

```

```python
dataset_limpo = dataset.dropna(axis=1)
display(dataset_limpo.dtypes)
```

```python
print(dataset_limpo['Positive'].head(10))
```

```python
# 1.Preparando os dados
eixo_positive = dataset_limpo['Positive']/100
eixo_negative = dataset_limpo['Negative']/100
eixo_recomend = dataset_limpo['Recommendations']/100

# 2. Criando o gráfico
plt.figure(figsize=(12,5))

# PRIMEIRO GRÁFICO
plt.subplot(1,2,1)
plt.scatter(eixo_positive, eixo_negative, color='green')
#plt.title('Avaliações positivas x Recomendações')
plt.xlabel('Avaliações positivas(em centenas de pessoas)')
plt.ylabel('Recomendações(em centenas de pessoas)')
plt.grid(True)
#plt.xticks(rotation=45)

# SEGUNDO GRÁFICO
plt.subplot(1,2,2)
plt.scatter(eixo_negative,eixo_recomend, color='red')
plt.xlabel('Avaliações negativas(em centenas de pessoas)')
plt.ylabel('Recomendações(em centenas de pessoas)')
plt.grid(True)
#plt.xticks(rotation=45)

# 3. Exibindo o gráfico
plt.tight_layout()
plt.show()
```

```python
dataset_numerico = dataset_limpo.copy()
dataset_numerico = dataset_numerico.select_dtypes(include=np.number)
corr_matrix = dataset_numerico.corr()
plt.figure(figsize=(16,12))

sns.heatmap(data=corr_matrix,annot=False, cmap='coolwarm')
plt.title('Heatmap de correlação das variáveis')
plt.show()
```
